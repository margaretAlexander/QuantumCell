```python
cell_membership = pd.DataFrame()

for size in metacell_sizes:
    cells = pd.read_csv(OUTPUT_DIR  + dataset_name + "_" + str(size) + "_cell_membership.csv", index_col = 0)
    cell_membership[str(size)] = cells['metacell_name']

cell_membership
```

```python
cell_membership.to_csv(OUTPUT_DIR + 'cell_membership.csv')
```
